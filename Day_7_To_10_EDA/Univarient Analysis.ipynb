```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=sns.load_dataset('titanic')
data.head()
```

```python
# catagorycal data
```

```python
fig=plt.figure(figsize=(10,6))
ax1=fig.add_subplot(121)
sns.countplot(data=data,x='survived',ax=ax1)

ax2=fig.add_subplot(122)
sns.countplot(data=data,x='pclass',hue='survived',ax=ax2)
plt.show()
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2)

data['survived'].value_counts().plot(kind='pie',autopct="%1.2f%%",labels=data['survived'].value_counts().index,ax=ax[0,0])
ax[0,0].set_title("survived pie",fontdict={'color':'red'})

data['pclass'].value_counts().plot(kind='pie',autopct="%1.2f%%",labels=data['pclass'].value_counts().index,ax=ax[0,1])
ax[0,1].set_title("pclass pie",fontdict={'color':'red'})

data['sex'].value_counts().plot(kind='pie',autopct="%1.2f%%",labels=data['sex'].value_counts().index,ax=ax[1,0])
ax[1,0].set_title("sex pie",fontdict={'color':'red'})

data['embarked'].value_counts().plot(kind='pie',autopct="%1.2f%%",labels=data['embarked'].value_counts().index,ax=ax[1,1])
ax[1,1].set_title("embarked pie",fontdict={'color':'red'})

plt.show()
```

```python
# numerical analysis
```

```python
fig,ax=plt.subplots(nrows=1,ncols=2,figsize=(10,5))

data['age'].plot(kind='hist',ax=ax[0])
ax[0].set_title("age histplot")

data['fare'].plot(kind='hist',ax=ax[1],log=True)
ax[1].set_title("fare histplot")

plt.show()
```

```python
sns.histplot(data=data,x='age',kde=True,element='step')
sns.rugplot(data=data,x='age',color='green')
plt.show()
```

```python
sns.boxplot(data=data,x='fare')
plt.show()
```

```python
# donut chart
counts = data['survived'].value_counts()
plt.pie(
    counts,
    labels=counts.index,
    autopct='%1.1f%%',
    wedgeprops={'width':0.2} )
plt.title("Donut Chart")
plt.show()
```
