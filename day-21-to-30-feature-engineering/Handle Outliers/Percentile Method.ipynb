```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv(r'weight-height.csv')
data.head()
```

```python
data.shape
```

```python
data.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(x=data['Height'],element='step',kde=True)
plt.show()
```

```python
data['Height'].skew()
```

```python
sns.boxplot(x=data['Height'])
plt.show()
```

```python
upper=data['Height'].quantile(0.99)
lower=data['Height'].quantile(0.01)
```

```python
upper,lower
```

```python
new_data=data[(data['Height']>=lower) & (data['Height']<=upper)]
```

```python
new_data.shape
```

```python
data['Height'].describe()
```

```python
new_data['Height'].describe()
```

```python
# Capping --> Winsorization
data['Height']=np.clip(data['Height'], lower, upper)
```

```python
data['Height'].describe()
```

```python
sns.histplot(data['Height'],element='step',kde=True)
plt.show()
```

```python
sns.boxplot(x=data['Height'])
plt.show()
```
