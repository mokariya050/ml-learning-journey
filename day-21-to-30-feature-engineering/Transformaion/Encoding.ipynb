```python

df = pd.DataFrame({
    "City":["Delhi","Mumbai","Delhi","Chennai","Delhi","Mumbai"]
})

freq = df["City"].value_counts()

df["City_encoded"] = df["City"].map(freq)

print(df)
```

```python
# Target Encoding

df = pd.DataFrame({
    "City":["Delhi","Delhi","Mumbai","Mumbai","Chennai"],
    "Purchased":[1,0,0,0,1]
})

mean = df.groupby("City")["Purchased"].mean()

df["City_Encoded"] = df["City"].map(mean)

print(df)
```
