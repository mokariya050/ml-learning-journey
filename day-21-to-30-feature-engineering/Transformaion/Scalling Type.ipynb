```python
    def c_education(data):
        text=data.strip()
        if text=="Graduate":
            return 1
        else:
            return 0
    def c_employee(data):
        text=data.strip()
        if text=="Yes":
            return 1
        else:
            return 0
```

```python
data['education']=data['education'].apply(c_education)
data['self_employed']=data['self_employed'].apply(c_employee)
```

```python
data.head()
```

```python
data['loan_status']=data['loan_status'].str.strip().map({"Approved":1,'Rejected':0})
```

```python
data.head()
```

```python
X_train,X_test,y_train,y_test=train_test_split(data.drop(columns=['loan_status']),data['loan_status'],test_size=0.33,random_state=42)
```

```python
lr=LogisticRegression()
lr.fit(X_train,y_train)

y_pred1=lr.predict(X_test)
score1=accuracy_score(y_test,y_pred1)
score1
```

```python
scaler1=StandardScaler()
X_train1=scaler1.fit_transform(X_train)
X_test1=scaler1.transform(X_test)

scaler2=MinMaxScaler()
X_train2=scaler2.fit_transform(X_train)
X_test2=scaler2.transform(X_test)

scaler3=RobustScaler()
X_train3=scaler3.fit_transform(X_train)
X_test3=scaler3.transform(X_test)

scaler4=MaxAbsScaler()
X_train4=scaler4.fit_transform(X_train)
X_test4=scaler4.transform(X_test)
```

```python
# standard scaler
lr1=LogisticRegression()
lr1.fit(X_train1,y_train)

y_pred1=lr1.predict(X_test1)
score1=accuracy_score(y_test,y_pred1)
score1
```

```python
# minmax scaler
lr2=LogisticRegression()
lr2.fit(X_train2,y_train)

y_pred2=lr2.predict(X_test2)
score2=accuracy_score(y_test,y_pred2)
score2
```

```python
# robust scaler
lr3=LogisticRegression()
lr3.fit(X_train3,y_train)

y_pred3=lr3.predict(X_test3)
score3=accuracy_score(y_test,y_pred3)
score3
```

```python
# Maxabs scaler
lr4=LogisticRegression()
lr4.fit(X_train4,y_train)

y_pred4=lr4.predict(X_test4)
score4=accuracy_score(y_test,y_pred4)
score4
```
