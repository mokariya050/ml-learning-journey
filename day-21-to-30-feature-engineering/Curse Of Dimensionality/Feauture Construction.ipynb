```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('train.csv',usecols=['Survived','Pclass','Age','SibSp','Parch','Fare'])
data.head()
```

```python
data.shape
```

```python
data.dropna(subset=['Age'],inplace=True)
```

```python
data.isna().sum()
```

```python
from sklearn.model_selection import train_test_split,cross_val_score

X_train,X_test,y_train,y_test=train_test_split(data.drop(columns=['Survived']),data['Survived'],test_size=0.2,random_state=42)
X_train.shape,X_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
lr=LogisticRegression()
lr.fit(X_train,y_train)

y_pred=lr.predict(X_test)

score=accuracy_score(y_test,y_pred)
score
```

```python
np.mean(cross_val_score(LogisticRegression(),data.drop(columns=['Survived']),data['Survived'],scoring='accuracy',cv=10))
```

```python
# Feature Contruction

data['Family']=data['SibSp']+data['Parch']+1
```

```python
data.head()
```

```python
data.drop(columns=['SibSp','Parch'],inplace=True)
```

```python
data['Family']=np.where(data['Family']>1,1,0)
```

```python
data.head()
```

```python
X_train,X_test,y_train,y_test=train_test_split(data.drop(columns=['Survived']),data['Survived'],test_size=0.2,random_state=42)
X_train.shape,X_test.shape
```

```python
lr=LogisticRegression()
lr.fit(X_train,y_train)

y_pred=lr.predict(X_test)

score=accuracy_score(y_test,y_pred)
score
```

```python
np.mean(cross_val_score(LogisticRegression(),data.drop(columns=['Survived']),data['Survived'],scoring='accuracy',cv=10))
```
