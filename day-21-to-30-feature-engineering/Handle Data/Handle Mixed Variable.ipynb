```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('titanic.csv')
data.head()
```

```python
data['number'].unique()
```

```python
data['numeric_number']=pd.to_numeric(data['number'],errors='coerce',downcast='integer')
```

```python
data.head()
```

```python
data.info()
```

```python
data['cat_number']=np.where(data['numeric_number'].isna(),data['number'],np.nan)
```

```python
data.head()
```

```python
data['cabin_num'] = data['Cabin'].str.extract('(\d+)') # captures numerical part
data['cabin_cat'] = data['Cabin'].str[0] # captures the first letter

data.head()
```

```python
import matplotlib.pyplot as plt
data['cabin_cat'].value_counts().plot(kind='bar')
plt.show()
```
