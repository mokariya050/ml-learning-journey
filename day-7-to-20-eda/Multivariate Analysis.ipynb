```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
tips.head()
```

```python
titanic=sns.load_dataset('titanic')
titanic.head()
```

```python
flight=sns.load_dataset('flights')
flight.head()
```

```python
iris=sns.load_dataset('iris')
iris.head()
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',size='size',style='day')
plt.title('total_bill and tips scatter plot')
plt.show()
```

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
plt.show()
```

```python
sns.jointplot(data=iris,x='petal_length',y='sepal_width',hue='species')
plt.show()
```

```python
g=sns.JointGrid(data=iris,x='sepal_length',y='petal_length')
g.plot(sns.scatterplot,sns.boxplot)
plt.show()
```

```python
plt.hexbin(tips['total_bill'],tips['tip'],cmap='viridis',mincnt=1, gridsize=20)
plt.colorbar()
plt.show()
```

```python
g=sns.JointGrid(data=iris,x='sepal_length',y='petal_length')

g.plot_joint(sns.scatterplot)

plt.histplot(iris['sepal_length'],ax=g.ax_marg_x)
sns.kdeplot(y=iris['petal_length'],ax=g.ax_marg_y)

plt.show()
```

```python
sns.boxplot(data=titanic,x='sex',y='age',hue='survived')
plt.show()
```

```python
sns.histplot(data=titanic,x='age',kde=True,element='step')
sns.histplot()
plt.show()
```

```python
sns.kdeplot(data=titanic,x='age',hue='survived')
plt.show()
```

```python
grid=pd.crosstab(titanic['survived'],titanic['pclass'])
```

```python
sns.heatmap(grid,annot=True)
plt.show()
```

```python
(titanic.groupby('pclass')['survived'].mean()*100).plot(kind='bar')
plt.show()
```

```python
titanic.groupby('pclass')['survived'].mean()
```

```python
temp=pd.crosstab(titanic['sibsp'],titanic['survived'])
sns.clustermap(temp.iloc[2:,:])
plt.show()
```

```python
sns.pairplot(data=iris,hue='species')
plt.show()
```

```python
temp=flight.groupby('year')['passengers'].sum().reset_index()
temp.head()
```

```python
sns.lineplot(x=temp['year'],y=temp['passengers'])
plt.show()
```

```python
sns.barplot(data=titanic,x='pclass',y='age',errorbar=None,hue='sex')
plt.show()
```
